# **PART ONE: UNDERSTANDING FEW-SHOT PROMPTING**

## **2. What is prompting?**

Prompting is the process of giving an AI model instructions or information to guide it towards producing a particular response.

For example, imagine you want an AI to classify customer support messages into categories.

You give it this instruction:

```
Classify the following customer message as either:
COMPLAINT, BILLING, TECHNICAL SUPPORT, or GENERAL ENQUIRY.

Customer message:
I was charged twice for my subscription.
```

The AI is expected to understand the instruction and identify the correct category.

The instruction you provide is called a prompt.

## **3. What is zero-shot prompting?**

Zero-shot prompting is when you ask an AI to perform a task without providing examples of how the task should be completed.

You provide the instruction, and the AI attempts to perform the task based on what it already understands.

**Example:**

```
Classify each customer message into one of these categories:

1. COMPLAINT
2. BILLING
3. TECHNICAL SUPPORT
4. GENERAL ENQUIRY

Customer message:
I cannot log into my account.

Return only the category.
```

Expected output:

```
TECHNICAL SUPPORT
```

Notice that the AI received instructions but was not shown any previous examples.

**Teaching explanation:**

Zero-shot prompting is like asking a student to solve a question after explaining the instructions but without showing the student a worked example.

The student must understand the instructions and apply their existing knowledge.

### **Advantages of zero-shot prompting**

- It is simple to write.
- It requires fewer instructions.
- It does not require preparing examples.
- It is useful for straightforward tasks.

### **Disadvantages of zero-shot prompting**

- The AI may misunderstand the intended categories.
- It may interpret ambiguous messages differently from what you expect.
- It may not follow a particular classification style.
- Its output may be inconsistent when instructions are unclear.

## **4. What is few-shot prompting?**

Few-shot prompting is a prompting technique in which you provide an AI with a small number of examples before asking it to perform a new task.

These examples demonstrate the expected pattern, format, or decision-making approach.

Instead of simply telling the AI what to do, you show it how the task should be performed.

**Example:**

```
Classify customer messages into these categories:

COMPLAINT
BILLING
TECHNICAL SUPPORT
GENERAL ENQUIRY

Examples:

Message: I was charged twice for my subscription.
Category: BILLING

Message: Your customer service team has been very unhelpful.
Category: COMPLAINT

Message: I cannot access my account.
Category: TECHNICAL SUPPORT

Now classify this message:

Message: I want to know when my subscription expires.
Category:
```

Expected output:

```
GENERAL ENQUIRY
```

The AI has received three examples before the new question. That is why this is called three-shot prompting.

### **Important terms**

| Term       | Meaning                                  |
| ---------- | ---------------------------------------- |
| Zero-shot  | No examples are provided.                |
| One-shot   | One example is provided.                 |
| Few-shot   | A small number of examples are provided. |
| Three-shot | Exactly three examples are provided.     |

**Important:** Few-shot prompting does not mean that the AI is being retrained. You are providing examples in the prompt to guide its response.

## **5. Zero-shot versus few-shot prompting**

| Feature      | Zero-shot             | Few-shot                                               |
| ------------ | --------------------- | ------------------------------------------------------ |
| Instructions | Provided              | Provided                                               |
| Examples     | None                  | A few                                                  |
| Guidance     | Based on instructions | Based on instructions and examples                     |
| Preparation  | Easier                | Requires example selection                             |
| Consistency  | May vary              | Can improve when examples are suitable                 |
| Best use     | Straightforward tasks | Tasks requiring a particular pattern or interpretation |

# **PART TWO: WHEN DO EXAMPLES HELP?**

## **6. Why do examples help AI?**

Examples help an AI understand what you expect from it.

Instructions tell the AI what task to perform, while examples demonstrate how you want that task performed.

Consider this instruction:

```
Classify customer messages into appropriate categories.
```

The instruction is not very specific.

What does "appropriate categories" mean?

Should a payment complaint be classified as BILLING or COMPLAINT?

Should a customer asking about payment methods be classified as BILLING or GENERAL ENQUIRY?

Examples help establish the intended interpretation.

### **Five important ways examples help**

**1. They clarify instructions**

Examples demonstrate what a category means in practice.

**2. They establish patterns**

Examples show the relationship between an input and its expected output.

**3. They reduce ambiguity**

They help communicate how borderline cases should be handled.

**4. They establish output format**

If you want the AI to return only a category, your examples can demonstrate that format.

**5. They improve consistency**

When examples are clear and relevant, they can encourage the AI to classify similar messages using the same approach.

## **7. Situations where examples are particularly useful**

### **Situation 1: Ticket classification**

An AI needs to separate customer messages into different support categories.

Examples demonstrate how different messages should be labelled.

### **Situation 2: Sentiment classification**

An AI needs to identify whether a message is positive, negative, or neutral.

Examples show what each sentiment category looks like.

### **Situation 3: Text formatting**

An AI needs to convert unstructured information into a particular format.

Examples demonstrate the expected structure.

### **Situation 4: Identifying spam**

An AI needs to distinguish promotional or suspicious messages from legitimate messages.

Examples demonstrate the characteristics of each category.

### **Situation 5: Extracting information**

An AI needs to identify names, dates, amounts, or other details from messages.

Examples demonstrate which information to extract and how to present it.

## **8. When might examples not help?**

Examples do not automatically improve every AI response.

They may be unhelpful when:

1. The examples are unrelated to the task.
2. The examples contain incorrect labels.
3. The examples contradict one another.
4. The examples are too similar to demonstrate different categories.
5. The examples encourage an incorrect interpretation.

For example:

```
Message: I was charged twice.
Category: TECHNICAL SUPPORT

Message: I cannot log into my account.
Category: BILLING

Message: My payment failed.
Category: COMPLAINT
```

These examples have confusing labels.

The AI may follow the incorrect patterns demonstrated in the prompt.

**Key lesson:** The quality of the examples matters more than simply increasing their number.

# **PART THREE: EXAMPLE SELECTION**

## **9. What is example selection?**

Example selection is the process of choosing suitable examples to include in a few-shot prompt.

It is not enough to provide any three examples. You must choose examples that demonstrate the task correctly.

For example, if you want an AI to classify customer support tickets, your examples should represent the types of messages it will actually encounter.

## **10. Characteristics of good examples**

A good example should have the following characteristics.

### **1. Relevance**

The example should relate directly to the task.

If you are classifying customer complaints, use customer messages rather than unrelated academic questions.

### **2. Correctness**

The example must have the correct category.

Incorrect labels can teach the AI the wrong pattern.

### **3. Clarity**

The example should be easy to understand.

Avoid messages that are unnecessarily complicated.

### **4. Diversity**

Your examples should represent different categories.

For instance, selecting three BILLING examples may not demonstrate how to distinguish BILLING from COMPLAINT or TECHNICAL SUPPORT.

### **5. Consistency**

Examples should use the same classification rules.

Do not label similar situations differently without explaining why.

## **11. Good versus poor example selection**

Consider the task of classifying customer support tickets into:

- BILLING
- COMPLAINT
- TECHNICAL SUPPORT
- GENERAL ENQUIRY

**Good example selection**

Suitable

| Customer message                        | Correct category  |
| --------------------------------------- | ----------------- |
| I was charged twice.                    | BILLING           |
| I cannot log into my account.           | TECHNICAL SUPPORT |
| Your delivery service is disappointing. | COMPLAINT         |

Why is this good?

The examples are relevant, correctly labelled, and demonstrate different categories.

**Poor example selection**

Unsuitable

| Customer message     | Assigned category |
| -------------------- | ----------------- |
| I was charged twice. | TECHNICAL SUPPORT |
| I cannot log in.     | BILLING           |
| My payment failed.   | COMPLAINT         |

Why is this poor?

The labels are incorrect, making the demonstrated pattern unreliable.

## **12. How to select good examples**

Use this simple five-step method.

**Step 1: Understand the task.**

Know exactly what you want the AI to do.

Example: Classify customer support messages.

**Step 2: Identify the possible categories.**

Example:

- BILLING
- COMPLAINT
- TECHNICAL SUPPORT
- GENERAL ENQUIRY

**Step 3: Collect representative messages.**

Find messages that clearly demonstrate each category.

**Step 4: Check the labels.**

Make sure every message is assigned to the correct category.

**Step 5: Select a balanced and useful set.**

Choose examples that demonstrate meaningful differences between categories.

# **PART FOUR: PRACTICAL TICKET CLASSIFICATION**

## **13. What is ticket classification?**

Ticket classification is the process of assigning customer support messages to predefined categories.

Businesses receive many messages from customers. These messages may involve payments, complaints, technical problems, or general questions.

Classifying them helps direct messages to the appropriate department.

For this lesson, we will use four categories:

| Category          | Meaning                                    | Example                 |
| ----------------- | ------------------------------------------ | ----------------------- |
| BILLING           | Payment, invoice, or subscription problems | I was charged twice.    |
| COMPLAINT         | Dissatisfaction with a product or service  | Your service is poor.   |
| TECHNICAL SUPPORT | Technical difficulties                     | I cannot log in.        |
| GENERAL ENQUIRY   | Requests for general information           | What time do you close? |

## **14. Understanding the difference between BILLING and COMPLAINT**

This distinction is particularly important.

Consider these two messages:

**Message A:**

```
I was charged twice for my subscription.
```

Correct category: BILLING.

The main issue is an incorrect payment.

**Message B:**

```
I am disappointed with how your company handles payments.
```

Correct category: COMPLAINT.

The customer is expressing dissatisfaction.

The messages are related to payments, but their main purposes differ.

This is where examples can help establish classification rules.

## **15. Practical demonstration: Zero-shot classification**

First, we will classify customer tickets without examples.

Use this prompt:

```
You are a customer support ticket classifier.

Classify each customer message into exactly one of
these categories:

BILLING
COMPLAINT
TECHNICAL SUPPORT
GENERAL ENQUIRY

Return only the category.

Customer message:
I have been trying to reset my password, but the
system keeps showing an error.
```

Expected output:

```
TECHNICAL SUPPORT
```

Notice that no examples were provided.

## **16. Practical demonstration: Three-shot classification**

Now we will give the AI three carefully selected examples.

```
You are a customer support ticket classifier.

Classify each customer message into exactly one of
these categories:

BILLING
COMPLAINT
TECHNICAL SUPPORT
GENERAL ENQUIRY

Examples:

Message: I was charged twice for my subscription.
Category: BILLING

Message: Your delivery service is very disappointing.
Category: COMPLAINT

Message: I cannot log into my account.
Category: TECHNICAL SUPPORT

Now classify this new message:

Message: I have been trying to reset my password,
but the system keeps showing an error.

Return only the category.
```

Expected output:

```
TECHNICAL SUPPORT
```

The AI has now seen three examples that demonstrate the intended categories.

# **PART FIVE: HANDS-ON PYTHON PRACTICE**

## **18. Preparing the ticket dataset**

We can represent our customer messages using Python.

```
tickets = [    {        "message": "I was charged twice this month.",        "expected": "BILLING"    },    {        "message": "Your customer service is terrible.",        "expected": "COMPLAINT"    },    {        "message": "My account keeps showing an error when I log in.",        "expected": "TECHNICAL SUPPORT"    },    {        "message": "What time does your office close?",        "expected": "GENERAL ENQUIRY"    },    {        "message": "I have not received my payment receipt.",        "expected": "BILLING"    },    {        "message": "I am unhappy with how your staff treated me.",        "expected": "COMPLAINT"    },    {        "message": "The application keeps crashing.",        "expected": "TECHNICAL SUPPORT"    },    {        "message": "Do you operate on weekends?",        "expected": "GENERAL ENQUIRY"    },    {        "message": "My subscription payment was deducted twice.",        "expected": "BILLING"
```

Each dictionary contains two important pieces of information:

- `message`: The customer's message.
- `expected`: The correct category.

## **19. Creating the zero-shot prompt**

```
def zero_shot_prompt(message):    prompt = f"""You are a customer support ticket classifier.Classify the customer message into exactly oneof these categories:BILLINGCOMPLAINTTECHNICAL SUPPORTGENERAL ENQUIRYReturn only the category.Customer message:{message}"""    return prompt
```

The function accepts a customer message and creates a prompt without examples.

## **20. Creating the three-shot prompt**

```
def three_shot_prompt(message):    prompt = f"""You are a customer support ticket classifier.Classify customer messages into exactly oneof these categories:BILLINGCOMPLAINTTECHNICAL SUPPORTGENERAL ENQUIRYExamples:Message: I was charged twice for my subscription.Category: BILLINGMessage: Your customer service is very poor.Category: COMPLAINTMessage: I cannot access my account.Category: TECHNICAL SUPPORTNow classify this message:Customer message:{message}Return only the category."""    return prompt
```

This function performs the same task but includes three examples.

**Teaching point:** Both functions accept the same input. The important difference is that one includes examples and the other does not.

## **21. Viewing the prompts**

```
sample_ticket = "My application keeps crashing."print("ZERO-SHOT PROMPT")print(zero_shot_prompt(sample_ticket))print("\nTHREE-SHOT PROMPT")print(three_shot_prompt(sample_ticket))
```

This allows us to inspect the actual prompts before sending them to an AI model.

## **22. Testing the prompts**

The following is a demonstration of how the test results can be recorded.

For an actual experiment, you would submit each generated prompt to the same AI model, record the returned category, and use those predictions in the comparison.

```
zero_shot_predictions = [    "BILLING",    "COMPLAINT",    "TECHNICAL SUPPORT",    "GENERAL ENQUIRY",    "BILLING",    "COMPLAINT",    "TECHNICAL SUPPORT",    "GENERAL ENQUIRY",    "BILLING",    "GENERAL ENQUIRY"]three_shot_predictions = [    "BILLING",    "COMPLAINT",    "TECHNICAL SUPPORT",    "GENERAL ENQUIRY",    "BILLING",    "COMPLAINT",    "TECHNICAL SUPPORT",    "GENERAL ENQUIRY",    "BILLING",    "GENERAL ENQUIRY"]
```

These are illustrative predictions, not measured results from a live AI model.

## **23. Calculating accuracy**

Accuracy tells us the percentage of predictions that are correct.

The formula is:

\\[ Accuracy = \frac{\text{Correct Predictions}}{\text{Total Predictions}}\times100 \\]

Python implementation:

```
def calculate_accuracy(expected, predicted):    correct = 0    for actual, prediction in zip(expected, predicted):        if actual == prediction:            correct += 1    accuracy = (correct / len(expected)) * 100    return accuracy
```

Now calculate the accuracy of both methods:

```
expected_categories = [    ticket["expected"] for ticket in tickets]zero_accuracy = calculate_accuracy(    expected_categories,    zero_shot_predictions)three_accuracy = calculate_accuracy(    expected_categories,    three_shot_predictions)print("Zero-shot accuracy:", zero_accuracy, "%")print("Three-shot accuracy:", three_accuracy, "%")
```

With the illustrative predictions above, both methods achieve 100%.

This is intentional: it demonstrates that few-shot prompting does not guarantee higher accuracy in every test.

## **24. Comparing the results**

```
if three_accuracy > zero_accuracy:    print("Three-shot prompting performed better.")elif zero_accuracy > three_accuracy:    print("Zero-shot prompting performed better.")else:    print("Both methods achieved the same accuracy.")
```

### **How to interpret the results**

- If three-shot accuracy is higher, the selected examples may have helped the AI classify the tickets more accurately.
- If zero-shot accuracy is higher, the examples may have been unnecessary, confusing, or poorly selected.
- If both are equal, the examples did not improve accuracy on that particular test set.

**Important experimental rule:** Use the same model, test messages, category definitions, and output requirements for both approaches. If the model produces variable responses, repeat the tests and report that variability.

# **PART EIGHT: COMMON MISTAKES**

## **30. Mistakes to avoid when using few-shot prompting**

| Mistake                                           | Why it is a problem                                          |
| ------------------------------------------------- | ------------------------------------------------------------ |
| Using irrelevant examples                         | They do not demonstrate the task.                            |
| Using incorrect labels                            | They teach the wrong classification pattern.                 |
| Choosing examples from only one category          | They may not demonstrate the differences between categories. |
| Using contradictory examples                      | They create confusion.                                       |
| Changing instructions between tests               | The comparison becomes unfair.                               |
| Testing with different messages                   | The comparison becomes unreliable.                           |
| Assuming more examples always mean better results | Example quality matters, not just quantity.                  |
| Claiming results without testing                  | It produces unsupported conclusions.                         |
